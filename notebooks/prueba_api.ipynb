```python
import json
from pathlib import Path 
from datetime import datetime

def ndjson_transform(json_file, staged_file):

    with json_file.open("r", encoding='utf-8') as file:
        data = json.load(file)

    with staged_file.open("w", encoding="utf-8") as staged_file:
        for coin in data:
            coin['snapshot_date'] = datetime.now().strftime('%Y-%m-%d')
            json_line = json.dumps(coin)
            staged_file.write(json_line)
            staged_file.write("\n")
        

ndjson_transform(Path.cwd().parent / "data" / "raw" /  r"market_snapshot_20260802_172137.json", Path.cwd().parent / "data" /r"prueba.ndjson")    
```
